```python
import pandas as pd 
import numpy as np
from sklearn.model_selection import train_test_split
import os 
import joblib
df = pd.read_csv('../Data/enhanced_collision.csv')
df.info()
```

```python
X = df.drop(['collision_year','collision_severity'],axis=1)
y= df['collision_severity']
X_train,X_test,Y_train,Y_test=train_test_split(X,y,test_size=0.2,stratify=y,random_state=42)
```

```python
print(X_train)
```

```python
print(X_test)
```

```python
from sklearn.preprocessing import StandardScaler
cols_toStandarize = ['longitude','latitude','speed_limit','number_of_vehicles','hour','month','number_of_casualties']
scaler = StandardScaler()
X_train[cols_toStandarize] = scaler.fit_transform(X_train[cols_toStandarize])
X_test[cols_toStandarize] = scaler.transform(X_test[cols_toStandarize])

```

```python
label_map = {1:0,2:1,3:2}
Y_train_mapped = Y_train.map(label_map) if hasattr(Y_train,'map') else pd.Series(Y_train).map(label_map)
Y_test_mapped = Y_test.map(label_map) if hasattr(Y_test,'map') else pd.Series(Y_test).map(label_map)
```

```python
os.makedirs('../Models', exist_ok=True)
joblib.dump(scaler, '../Models/scaler.pkl')
X_train.to_csv('../Data/X_train.csv', index=False)
X_test.to_csv('../Data/X_test.csv', index=False)
```

```python
print(Y_train_mapped.unique())
print(Y_test_mapped.unique())
```

```python
np.save('../Data/X_train.npy',X_train)
np.save('../Data/X_test.npy',X_test)
np.save('../Data/Y_train.npy',Y_train_mapped)
np.save('../Data/Y_test.npy',Y_test_mapped)

```
